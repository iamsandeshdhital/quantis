# QUANTIS Demo

QUANTIS — QUantum Advanced Nanoscale Technology & Information Systems

```python
import sys
sys.path.insert(0, '../src')

import numpy as np
from qit.error_correction.surface_code import create_surface_code
from qit.algorithms.circuit_optimizer import QuantumCircuit, Gate, CircuitOptimizer
from qit.ml.quantum_kernel import create_quantum_kernel, benchmark_quantum_ml

print('QUANTIS loaded successfully!')
```

## 1. Surface Code

```python
code = create_surface_code(distance=3)
print(f'Distance: {code.distance}')
print(f'Data qubits: {code.num_data_qubits}')
print(f'Stabilizers: {len(code.stabilizers)}')

# Error simulation
result = code.simulate_error(error_rate=0.01, num_trials=100)
print(f'Success rate: {result["success_rate"]:.2%}')
```

## 2. Circuit Optimization

```python
circuit = QuantumCircuit(num_qubits=4, name='demo')
circuit.add_gate(Gate('H', [0]))
circuit.add_gate(Gate('X', [1]))
circuit.add_gate(Gate('H', [0]))
circuit.add_gate(Gate('CNOT', [0, 1]))
circuit.add_gate(Gate('CNOT', [0, 1]))

optimizer = CircuitOptimizer()
optimized = optimizer.optimize(circuit)
print(f'Original gates: {circuit.gate_count()}')
print(f'Optimized gates: {optimized.gate_count()}')
```

## 3. Quantum ML

```python
results = benchmark_quantum_ml(n_samples=30)
print(f'VQC accuracy: {results["vqc"]["accuracy"]:.2f}')
```
